```python
df = pd.read_csv('flood.csv')

print("Dataset Loaded Successfully")
df.head()
```

```python
print("Shape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
correlation_matrix = df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(15,10))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df['FloodProbability'],
    bins=30,
    kde=True
)

plt.title("Flood Probability Distribution")
plt.xlabel("Flood Probability")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(18,8))

sns.boxplot(data=df)

plt.xticks(rotation=90)

plt.title("Outlier Detection")
plt.show()
```

```python
corr_target = df.corr()['FloodProbability']

corr_target = corr_target.sort_values(ascending=False)

print(corr_target)
```

```python
top_features = corr_target.drop('FloodProbability').head(10)

print(top_features)
```

```python
average_values = df.mean()

average_values
```

```python
plt.figure(figsize=(15,6))

average_values.plot(
    kind='bar'
)

plt.title("Average Value of Each Feature")
plt.xlabel("Features")
plt.ylabel("Average Value")

plt.show()
```

```python
sample_df = df.sample(500, random_state=42)

sns.pairplot(
    sample_df[
        [
            'MonsoonIntensity',
            'TopographyDrainage',
            'RiverManagement',
            'FloodProbability'
        ]
    ]
)

plt.show()
```
